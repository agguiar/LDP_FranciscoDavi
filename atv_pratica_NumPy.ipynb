# Nível 1
**Passo 1 — Array de peças vendidas**

```python
import numpy as np

pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])

print(pecas_vendidas)
```

**Passo 2 — Faturamento diário**

```python
preco_medio = 50

faturamento = pecas_vendidas * preco_medio

print("Faturamento diário:")
print(faturamento)
```

**Passo 3 — Dias com mais de 200 peças**

```python
mascara = pecas_vendidas > 200

dias_pico = pecas_vendidas[mascara]

print("Máscara:")
print(mascara)

print("\nVendas acima de 200 peças:")
print(dias_pico)
```

# Nível 2
**Passo 1 — Matriz das três lojas**

```python
import numpy as np

vendas = np.array([
    [200, 220, np.nan, 250],
    [150, 180, 160, 190],
    [300, 310, 290, 330]
])

print(vendas)
```

**Passo 2 — Total vendido por loja**

```python
total_por_loja = np.nansum(vendas, axis=1)

print("Total vendido por loja:")
print(total_por_loja)
```

**Passo 3 — Média ignorando nan**

```python
media_geral = np.nanmean(vendas)

print("Média diária geral da rede:")
print(media_geral)
```

**Passo 4 — Meta atingida ou abaixo**

```python
resultado_meta = np.where(
    vendas >= 200,
    "Meta Atingida",
    "Abaixo"
)

print(resultado_meta)
```

# Nível 3
**Passo 1 — Gerar 50 vendas aleatórias**

```python
import numpy as np

rng = np.random.default_rng(42)

departamentos = np.array([
    "Eletrônicos",
    "Roupas",
    "Casa"
])

vendas = rng.choice(departamentos, size=50)

print("Vendas geradas:")
print(vendas)
```

**Passo 2 — Contagem por departamento**

```python
categorias, contagens = np.unique(
    vendas,
    return_counts=True
)

print("Volume de vendas por departamento:")

for categoria, quantidade in zip(categorias, contagens):
    print(categoria, ":", quantidade)
```

**Passo 3 — Campanha campeã**

```python
faturamento_campanhas = np.array([
    12000,
    45000,
    23000,
    89000,
    31000
])

indice_campea = np.argmax(faturamento_campanhas)

print("Índice da campanha campeã:", indice_campea)
print("Faturamento:", faturamento_campanhas[indice_campea])
```

# Nível 2
**Passo 1 — reshape(4, 3)**

```python
visitas = np.array([
    1200, 1350, 1500,
    1600, 1700, 1800,
    1900, 2000, 2100,
    2200, 2300, 2400
])

trimestres = visitas.reshape(4, 3)

print("Visitas organizadas por trimestre:")
print(trimestres)
```

**Passo 2 — vstack()**

```python
primeiro_semestre = np.array([
    [1200, 1350, 1500],
    [1600, 1700, 1800]
])

segundo_semestre = np.array([
    [1900, 2000, 2100],
    [2200, 2300, 2400]
])

ano_completo = np.vstack([
    primeiro_semestre,
    segundo_semestre
])

print("Dados combinados:")
print(ano_completo)
```

**Passo 3 — Primeiro trimestre com .copy()**

```python
primeiro_trimestre = ano_completo[0].copy()

print("Primeiro trimestre:")
print(primeiro_trimestre)
```

**Passo 4 — ravel()**

```python
relatorio = ano_completo.ravel()

print("Array 1D:")
print(relatorio)
```

# Nível 3
**Passo 1 — Ranking com argsort()**

```python
notas = np.array([85, 92, 78, 95, 88])

indices = np.argsort(notas)

print("Índices em ordem crescente:")
print(indices)

print("Notas em ordem crescente:")
print(notas[indices])
```

**Passo 2 — np.newaxis e Broadcasting**

```python
precos = np.array([100, 200, 300])

precos_coluna = precos[:, np.newaxis]

print("Preços em coluna:")
print(precos_coluna)

print("Novo shape:", precos_coluna.shape)

descontos = np.array([0.10, 0.20, 0.30])

precos_com_desconto = precos_coluna * (1 - descontos)

print("Preços com descontos:")
print(precos_com_desconto)
```

**Passo 3 — Sistema com np.linalg.solve()**

```python
A = np.array([
    [2, 1],
    [1, -1]
])

b = np.array([
    500,
    100
])

solucao = np.linalg.solve(A, b)

print("Valor da placa: R$", round(solucao[0], 2))
print("Valor do sensor: R$", round(solucao[1], 2))
```
