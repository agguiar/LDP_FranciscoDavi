Exercício:  Análise de Desempenho de Clubes no Campeonato Brasileiro 2024

**Objetivo:**


*   Manipular dados de um campeonato de futebol.
*   Calcular e visualizar estatisticas de desempenho dos clubes.

Passo 1: Leitura da Planilha

Carregar dados do arquivo `brasileirao.xlsx` usando pandas.

```python
import pandas as pd

df = pd.read_excel('brasileirao.xlsx')

df.head()
```

Passo 2: Calcular o saldo de gols(SG)

Adicionar uma nova coluna chamada "Saldo de Gols" que será a diferença entre Gols Pró(GP) e Gols Contra(GC).

```python
df['Saldo de Gols'] = df['GP'] - df['GC']

df.head(20)
```

Passo 3: Calcular a pontuação

```python
df['Pontos'] = df['V'] * 3 + df['E']

df.head(20)
```

# Desafios Extras


**Desafio 1:** Melhor saldo de gols?

Exibir o clube com maior saldo.

**Desafio 2:** Maior quantidade de vitórias?

Ordenar os clubes pelo número de vitórias e exibir o top 5.

```python
melhor_saldo = df[df['Saldo de Gols'] == df['Saldo de Gols'].max()]
melhor_saldo
```

```python
top_vitorias = df.sort_values(by='V', ascending=False).head(5)
top_vitorias
```

```python
tabela_final = df.sort_values(by='Pontos', ascending=False)
tabela_final
```
