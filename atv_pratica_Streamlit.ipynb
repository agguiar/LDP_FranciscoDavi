```python
import streamlit as st
import pandas as pd

# Título
st.title("Dashboard de Vendas")

# Carregamento dos dados
@st.cache_data
def carregar_dados():
    df = pd.read_csv("vendas_brasil.csv")
    return df

df = carregar_dados()

# Painel lateral
st.sidebar.title("Filtros")

# Lista de categorias
lista_de_categorias = df["categoria"].unique()

# Filtro de categorias
categorias_selecionadas = st.sidebar.multiselect(
    "Selecione as Categorias",
    options=lista_de_categorias
)

# Aplicação do filtro
if categorias_selecionadas:
    df_filtrado = df[
        df["categoria"].isin(categorias_selecionadas)
    ].copy()
else:
    df_filtrado = df.copy()

# Conversão da coluna de data
df_filtrado["data"] = pd.to_datetime(
    df_filtrado["data"]
)

# Métricas
receita_calculada = df_filtrado["receita"].sum()

# Cada linha do arquivo é considerada um registro de venda
total_pedidos = len(df_filtrado)

# Duas colunas
col1, col2 = st.columns([1, 1])

with col1:
    st.metric(
        label="Receita Total",
        value=f"R$ {receita_calculada:,.2f}"
    )

with col2:
    st.metric(
        label="Total de Pedidos",
        value=total_pedidos
    )

# Abas
aba1, aba2 = st.tabs([
    "Evolução Mensal",
    "Tabela de Dados"
])

# Aba 1
with aba1:
    st.subheader("Evolução Mensal da Receita")

    df_filtrado["mes"] = (
        df_filtrado["data"]
        .dt.to_period("M")
        .astype(str)
    )

    dados_agrupados = (
        df_filtrado
        .groupby("mes")["receita"]
        .sum()
    )

    st.area_chart(dados_agrupados)

# Aba 2
with aba2:
    st.subheader("Tabela de Dados")

    st.dataframe(
        df_filtrado,
        use_container_width=True
    )

    csv = df_filtrado.to_csv(
        index=False
    ).encode("utf-8")

    st.download_button(
        label="Baixar dados em CSV",
        data=csv,
        file_name="vendas_filtradas.csv",
        mime="text/csv"
    )
```
